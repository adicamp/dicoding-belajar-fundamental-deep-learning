```python
# Import library
import pandas as pd
from google_play_scraper import Sort, reviews
import os
```

**Scraping Data Reviews**

```python
def scrape_reviews(app_id, lang='id', country='id'):
  """
  Function to scrape reviews from the Google Play Store and save them to a CSV file.
  """
  print(f'Starting review scraping for App ID: {app_id}...')

  # Scrape reviews using the google-play-scraper library
  results, _ = reviews(
    app_id,
    lang=lang,
    country=country,
    sort=Sort.MOST_RELEVANT,
    count=20000
  )

  # Convert the list of dictionaries into a Pandas DataFrame
  df = pd.DataFrame(results)

  # Ensure the storage directory exists
  directory = 'data'
  if not os.path.exists(directory):
    os.makedirs(directory)
    print(f"Directory '{directory}' created successfully.")

  # Generate a dynamic filename based on the app_id
  file_name = f"{app_id.split('.')[1]}_app_reviews.csv"
  file_path = os.path.join(directory, file_name)

  # Save the dataset to a CSV file
  df.to_csv(file_path, index=False)

  print(f"Successfully saved {len(df)} reviews to: {file_path}")
  return df
```

```python
# Scraping reviews for aplication SIGNAL–SAMSAT DIGITAL NASIONAL
app_id_target = 'app.signal.id'
df_reviews = scrape_reviews(app_id_target)
```
